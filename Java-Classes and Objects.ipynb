A class outlines the necessary components and how they interact with each other. 

Class is a **reference(Object) data types**

What are the relevant details of a savings account in a bank? How about these fields (**Attributes**):

## Constructor method with Parameters

In Java, parameters are placeholders that we can use to pass information to a method.

Since the constructor is a **method**, **we can include parameters to assign values to instance fields**.

**syntax**:

```java
//Declare Class
accessModifier class class_name {
    
    //Declare attributes
    accessModifier datatype var_name1;
    accessModifier datatype var_name2;
    accessModifier datatype var_name3;
    ...
    
    //Constructor method
    public Constructor_name(datatype param1, datatype param2,...){
        //parameter value assigned to the attributes
        var_name1 = param1;
        var_name2 = param2;
        ...
    }
}
```

**Example**:

If the Car constructor has a parameter: String carColor

```java
//Declare class
public class Car {
    
    //attributes
    public String color;
    
    //constructor method with parameter
    public Car (String carColor){
        color = carColor;
    }
    
    //main method 
    public static void main(String[] args) {
        
        //Create object with contructor method and assign color = red
        Car Tesla = new Car(red);
        System.out.println(Tesla.color); //Prints red
    }
}
```

### Multiple constructor method

A class **can have multiple constructors**. We can **differentiate them based on their parameters**. 

The signature helps the compiler to differentiate between different methods.

**Example**:

```java
//Declare Class
public class Car {

    //attributes
    public String color;
    public int mpg;
    public boolean isElectric;
    
    //constructor1 method 
    public Car(String carColor, int milesPerGallon) {
        color = carColor;
        mpg = milesPerGallon;
    }
    
    //constructor2 method
    public Car(boolean electricCar, int milesPerGallon) {
        isElectric = electricCar;
        mpg = milesPerGallon;
    }
}
```

1) The first constructor has two parameters: 
    * String carColor
    * int milesPerGallon.
2) The second one has these: 
    * boolean electricCar
    * int milesPerGallon.

The values will help the compiler to decide which constructor to use.

For example, **Car myCar = new Car(true, 40)** will be **created by the second constructor** because the arguments match the type and order of the second constructor’s signature.

### Without constructor method

When we **don’t define the constructor**, the Java compiler **creates a default constructor that assigns default values to an instance.**

**Default values can be created by assigning values to the instance fields during their declaration:**

**Example**:

```java
//Declare class
public class Car {

    //default attributes
    public String color = "red";
    public boolean isElectric = True;
    public int cupHolders = 2;
    
    //main method
    public static void main(String[] args) {
    
        //Create object
        Car Tesla = new Car();
        
        System.out.println(Tesla.color); // Prints: red 
        System.out.println(Tesla.isElectric); // Prints: true
        System.out.println(Tesla.cupHolders); // Prints: 2
    }
}
```

## Class Methods

Methods are repeatable, modular blocks of code used to accomplish specific tasks. 

We have the ability to **define our own methods that will take input, do something with it, and return the kind of output we want.**

To create a class method within the class:

**syntax**:

```java
//Declare Class
accessModifier class class_name {
    
    //Declare attributes
    accessModifier datatype var_name1;
    accessModifier datatype var_name2;
    accessModifier datatype var_name3;
    ...
    
    //Constructor method
    public Constructor_name(datatype param1, datatype param2,...){
        //parameter value assigned to the attributes
        var_name1 = param1;
        var_name2 = param2;
        ...
    }
    
    //class method
    visibilityModifier additionalModifiers return_datatype method_name(params...) {
    
    }
}
```

**visibilityModifier** can be:

|Visibility Modifiers|Descriptions|
|---|---|
|**public**|Methods declared as public can be called from everywhere, inside and outside of the object or the class.|
|**private**|Methods declared as private can only be called from inside the object or the class.|
|**package private**|Methods declared as package private can only be called from classes within the same package.|
|**protected**|Methods declared as protected can only be called from inside the class or from inside classes inherited from that class.|

**Additional Modifier** can be:

|Additional Modifiers|Descriptions|
|---|---|
|**final**|Attributes and methods cannot be overridden/modified|
|**static**|Attributes and methods belongs to the class, rather than an object|
|**abstract**|Can only be used in an abstract class, and can only be used on methods. The method does not have a body, for example abstract void run();. The body is provided by the subclass (inherited from).|
|**transient**|Attributes and methods are skipped when serializing the object containing them|
|**synchronized**|Methods can only be accessed by one thread at a time|
|**volatile**|The value of an attribute is not cached thread-locally, and is always read from the "main memory"|

**return datatype** can be:
|Return Datatypes|Descriptions|
|---|---|
|**int**| which stores whole numbers||
|**double**| which stores bigger whole numbers and decimal numbers, sufficient for storing 15 decimal digits|
| **boolean**| which stores true and false|
| **char**| which stores a single character/letter or ASCII values using single quotes '...'|
| **String**| which stores multiple characters using double quotes "..."|
| **byte**| Stores whole numbers from -128 to 127|
| **long**| Stores whole numbers from -9,223,372,036,854,775,808 to 9,223,372,036,854,775,807|
| **short**| Stores whole numbers from -32,768 to 32,767|
| **float**| Stores fractional numbers. Sufficient for storing 6 to 7 decimal digits|
|**void**|no output|

**Example**:

```java
//Declare Class
public class SavingsAccount {
  
    //Attributes
    int balance;
  
    //constructor method
    public SavingsAccount(int initialBalance){
    balance = initialBalance;
    }
    
    //Class method
    public void checkBalance(){
        System.out.println("Hello!");
        System.out.println("Your balance is " + balance);
    }
}
```

## Scope

A method is a task that an object of a class performs.

We mark the domain of this task using curly braces: {, and }. 

Everything inside the curly braces is part of the task. This domain is called the scope of a method.

**We can’t access variables that are declared inside a method in code that is outside the scope of that method.**

**Example**:

```java
//Declare class
public class Car {
    
    //Attributes
    String color;
    int milesDriven;

    //constructor method
    public Car(String carColor) {
    color = carColor;
    milesDriven = 0;         
    }
    
    //class method
    public void drive() {
        String message = "Miles driven: " + milesDriven;
        System.out.println(message);
    }
    
    //main method
    public static void main(String[] args){
        
        //Create Object
        Car myFastCar = new Car("red");
        
        //call method
        myFastCar.drive();
    }
}
```

The variable **message**, which is declared and initialized inside of drive() method, **cannot be used inside of main()!** 
\
It only exists within the scope of the drive() method.

However, **milesDriven**, which is declared at the top of the class, **can be used inside all methods in the class, since it is in the scope of the whole class.**

## Modify Attributes

Method can be used to modify the default value of the attribute

**Example**:

In the Savings account example, we want to implement 2 methods for depositing ad withdrawing.

```java
//Declare Class
public class SavingsAccount{

    //Attributes
    double balance;
    
    //Concstructor method
    public SavingsAccount(double startingBalance){
        balance = startingBalance;
    }
    
    //Class method for deposit
    public void deposit(double amountToDeposit){
        //Add amountToDeposit to the balance
        newBalance = balance + amountToDeposit;
        balance = newBalance;
    }

    //Class method for withdraw
    public void withdraw(double amountToWithdraw){
        //Subtract amountToWithdraw from the balance
        newBalance = balance + amountToWithdraw;
        balance = newBalance;
    }
    
    //main method
    public static void main(String[] args){
        
        //create object with initial balance = 500
        SavingsAccount myAccount = new SavingsAccount(500);
        
        //call attributes
        System.out.println(myAccount.balance); //print 500
        
        //call deposit method
        myAccount.deposit(100);
        
        //After deposit 100 into myAccount
        System.out.println(myAccount.balance); //print 600
        
        //call withdraw method
        myAccount.withdraw(200);
        
        //After withdraw 200 from myAccount
        System.out.println(myAccount.balance); //print 400
    }
}
```

## Return Statement

The **return** keyword is used to return a value from a method 

Instead of usng the keyword "void", we need to specify the datatype when declare the method.

**Example**:

```java
//Declare class
public class Car {
    
    //Attributes
    String color;
    int milesDriven;

    //constructor method
    public Car(String carColor) {
    color = carColor;
    milesDriven = 0;         
    }
    
    //class method
    public void drive() {
        String message = "Miles driven: " + milesDriven;
        System.out.println(message);
    }
    
    //class method 
    public int numberOfTires() {
        int tires = 4;
        
        // return statement
        return tires;
    }
    
    //main method
    public static void main(String[] args){
        
        //Create Object
        Car myFastCar = new Car("red");
        
        //assign return value of the method to a variable
        int numTires = myFastCar.numberOfTires; 
        
        System.out.println(numTires);
        
    }
}
```

We can also use method to return a object from the class too

**Example**:

```java
//Declare class
public class Car {
    
    //Attributes
    String color;
    int milesDriven;

    //constructor method
    public Car(String carColor) {
    color = carColor;
    milesDriven = 0;         
    }
    
    //class method
    public void drive() {
        String message = "Miles driven: " + milesDriven;
        System.out.println(message);
    }
    
    //class method 
    public int numberOfTires() {
        int tires = 4;
        
        // return statement
        return tires;
    }
}

//Declare 2nd Class
class CarLot {

    //Attributes
    Car carInLot;
    
    //Constructor method
    public CarLot(Car givenCar) {
        carInLot = givenCar;
    }
    
    //method (return Car Object)
    public Car returnACar() {
    
        // return Car object
        return carInLot;
    }
    
    //main method
    public static void main(String[] args) {
        
        //Create car object
        Car myCar = new Car("red", 70);
        System.out.println(myCar); 
        
        //Create carlot object
        CarLot myCarLot = new CarLot(myCar);
        System.out.println(myCarLot.returnACar());
    }
}
```

## toString method

When we print out Objects, it prints out the memory location of where the object is stored.

This doesn’t tell us anything about what the Store sells, the price, or the other instance fields we’ve defined. **We can add a toString() method to our classes that makes this printout more descriptive.**

When we define a toString() method for a class, we can **return a String that will print when we print the object:**

**Example**:

```java
//Declare Class
class Car {
    
    //Attributes
    String color;
    
    //constructor method
    public Car(String carColor) {
        color = carColor;
    }
    
    //toString method
    public String toString(){
        return "This is a " + color + " car!";
    }
    
    //main method
    public static void main(String[] args){
        Car myCar = new Car("red");
        System.out.println(myCar); // print This is a red car!
    }

}
```
