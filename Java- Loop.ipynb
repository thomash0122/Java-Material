## Loop

A loop is a programming tool that allows developers to **repeat the same block of code until some condition is met.**

Process: 
1) A starting condition is evaluated. 
2) If the starting condition is true, then the loop body is executed.
3) When the last line of the loop body is executed, the condition is re-evaluated.
4) Repeat 1.

This process continues until the condition is false (if the condition never becomes false, we can actually end up with an infinite loop!). If the starting condition is false, the loop never gets executed.

We employ loops to easily scale programs - saving time and minimizing mistakes.

There are 3 types of loop:
* **while loops**
* **for loops**
* **for-each loops**

## While-Loop

A while loop will **only run if the condition is true.** 

However, a while loop will **continue running the code over and over until the condition evaluates to false.**

**syntax**:

```java
while(logical_statement){
     //run this block if the logical_statement is true
}

//exist the while loop if the logical_statement is false
```

**Example**:

```java
// Importing the Random library
import java.util.Random;

  // Creating a random number generator
    Random randomGenerator = new Random();
    
    // Generate a number between 1 and 6
    int dieRoll = randomGenerator.nextInt(6) + 1;

    // Repeat while roll isn't 5
    while (dieRoll != 5){
      dieRoll = randomGenerator.nextInt(6) + 1;
      System.out.println(dieRoll);
    }
```

## incrementing

When looping through code, it’s common to use a counter variable. 

A counter (also known as an iterator) is a variable used in the conditional logic of the loop and (usually) **incremented in value during each iteration through the code.**

**Example**:

```java
int count = 1;

while (count<6){

    System.out.println("count: " + count);
    count++;
}
```

## For-Loop

Even though we can write while loops that accomplish the same task, for loops are useful because they help us remember to increment our counter — something that is easy to forget when we increment with a while loop.

**syntax**:

```java
for (int i = starting_value; i < ending_value; i++){
    //run this block when i < ending_value
    
}
```

**Example**:

```java
for (int i = 1; i < 6 ; i++){
    System.out.println("count: " + i);
}
```

## Iterating Over Arrays and ArrayLists

One common pattern we’ll encounter as a programmer is **traversing, or looping**, through a list of data and doing something with each item. 

In Java, that list would be an array or ArrayList and the loop could be a for loop.

**syntax**:

```java
// for Arrays
for (int i=0; i<Array_name.length; i++){
    //Operation on Array...
}

//for ArrayLists
for (int i=0; i<ArrayList_name.size(); i++){
    //Operation on ArrayList...
    
}
```

**Example**:

```java
import java.util.Arrays;

int[] numList = {1,2,3,4,5};

for (int i=0; i<numList.length; i++){
    // increment value of element value by 1
    numList[i] += 1;
}

System.out.println(Arrays.toString(numList));
```

```java
ArrayList <Integer> numArrayList = new ArrayList <Integer>();

for(int i=1; i<6; i++){
    numArrayList.add(i);
}

for (int i=0; i<numArrayList.size(); i++){
    // increment value of element value by 1
    int element = numArrayList.get(i);
    numArrayList.set(i, element+1);
}

System.out.println(numArrayList);
```

## Break and Continue

If we ever **want to exit a loop before it finishes all its iterations or want to skip one of the iterations**, we can use the **break** and **continue** keywords.

**Example**:

```java
for(int i=0; i<10; i++){
    System.out.println(i);
    
    // break the loop if i==4
    if(i==4){
    break;
    }
}
```

```java
int[] numbers = {1,2,3,4,5};

for (int i=0; i<numbers.length; i++){
    
    //skip if the element is even
    if(numbers[i] % 2 ==0){
        continue;
    }
    
    System.out.println(numbers[i]);
}
```

## For-Each Loops

For-each loops, which are also referred to as enhanced loops, **allow us to directly loop through each item in a list of items** (like an array or ArrayList) and perform some action with each item.

**syntax**:

```java
for (datatype element : ArrayOrArrayList){

    //execute block
}
```

**Example**:

```java
int[] numList = {1,2,3,4,5};

for (int num : numList){
    System.out.println(num);
}
```

We can read the : as “in” like this: for each num (which should be a int) in numList, print num.

```java
int[] numbers = {6, 29, 28, 33, 11, 100, 101, 43, 89};

public boolean isPrime(int number){
    if(number<=1){
        return false;
    } else if(number==2){
        return true;
    } else if(number>2){
        for(int i=2; i<number; i++)
            if(number%i==0){
                return false;
            }
    }
    return true;
}

for (int number:numbers){
    System.out.println(isPrime(number));
}
```
