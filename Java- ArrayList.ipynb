The ArrayList class uses **dynamic** arrays that are **resizable**, unlike traditional fixed arrays. **However, each element must still be of the same type**. Elements can be added or removed at any time, making the ArrayList more flexible.

To create mutable and dynamic lists, we can use Java’s ArrayList class. ArrayList allows us to:

* **Store object references as elements**
* **Store elements of the same type (just like arrays)**
* **Access elements by index (just like arrays)**
* **Add elements**
* **Remove elements**

## ArrayList Package

This class is imported with the java.util package.

```java
import java.util.ArrayList;
```

## Creating an Empty ArrayList

Unlike java array, in order to create a arraylist, **we must create an empty arraylist and then use arraylist method to add element into the arraylist**.

To create an empty ArrayList:

**syntax**:

```java
ArrayList<DataType> arrayListName = new ArrayList<DataType>();
```

DataType doesn't follow the standard naming. DataType can be as follow:
* **String**
* **Integer**
* **Double**
* **Boolean**
* **Can be leave as empty**

If leave as empty, then the ArrayList can store different values because we do not specify its type during initialization.

**Example**:

```java
ArrayList<Double> numList = new ArrayList<Double>();
```

## Adding an Item

ArrayList comes with an add() method which inserts an element into the structure. 

There are two ways we can use add():
* specify the index of where to add the item
* add the item at the end of the arraylist

**syntax**:

```java
ArrayListName.add(index_position, element_name);
```

**Example**:

```java
//add the item at the end of the arraylist

ArrayList<String> carShow = new ArrayList<String>();

carShow.add("ferrari");
// carShow now holds [ferrari]
carShow.add("thunderbird");
// carShow now holds [ferrari, thunderbird]
carShow.add("volkswagen");
// carShow now holds [ferrari, thunderbird, volkswagen]

System.out.println(carShow);
```

```java
//specify the index of where to add the item

carShow.add(2,"Toyota");
System.out.println(carShow);
```

## ArrayList Size

If we wanted to display the number of items in the ArrayList, we could find the size of it using the .size() method.

**syntax**:

```java
ArrayListName.size();
```

**Example**:

```java
//carShow: [ferrari, thunderbird, Toyota, volkswagen]

System.out.println(carShow.size());
```

## Accessing an Index

With arrays, we can use bracket notation to access a value at a particular index.

For ArrayLists, bracket notation won’t work. Instead, we use the .get() method to access an index.

**syntax**:

```java
ArrayListName.get(element_name);
```

**Example**:

```java
//carShow: [ferrari, thunderbird, Toyota, volkswagen]

System.out.println(carShow.get(1));
```

## Changing a Value

When we were using arrays, we could rewrite entries by using bracket notation to reassign values.

ArrayList has a slightly different way of doing this, using the .set() method.

**syntax**:

```java
ArrayListName.set(index_position, change_value);
```

**Example**:

```java
//carShow: [ferrari, thunderbird, Toyota, volkswagen]

carShow.set(0, "Lambo");
System.out.println(carShow);
```

## Removing an Item

ArrayLists allow us to remove an item by specifying the index to remove using the .remove method.

There are 2 ways to remove an item from ArrayList:
* **specify which element to remove**
* **specify which index position of the element to remove**

**syntax**:

```java
ArrayListName.remove(index_position or element_name);
```

**Example**:

```java
//carShow: [Lambo, thunderbird, Toyota, volkswagen]

carShow.remove(2);
System.out.println(carShow);
```

```java
//carShow: [Lambo, thunderbird, volkswagen]

carShow.remove("Lambo");
System.out.println(carShow);
```

## Getting an Item's Index

.indexOf() method gives the **index position of the given element.**

**syntax**:

```java
ArrayListName.indexOf(element_name);
```

**Example**:

```java
//carShow: [thunderbird, volkswagen]

System.out.println(carShow.indexOf("volkswagen"));
```
