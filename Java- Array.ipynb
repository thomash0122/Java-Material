## Java Array

In Java, an array is a **data structure** that can store **a fixed-size sequence of elements** of the **same data type**. 

An array is an **object in Java**, which means it **can be assigned to a variable**, **passed as a parameter to a method**, and **returned as a value from a method**.

## Comparsion between Python List and Java Array

|Python List|Java Array|
|---|---|
|**Dynamic**- meaning the size of a list can be modified |**Static**- meaning the size of a array is fixed|
|Store **multiple datatype**|Store **only one datatype**|
|**mutable**- can modify an element in the list|**mutable**- can modify an element in the array|
|**Have built-in method**|**Doesn't have built-in method**, need to pass the array to the methods in Arrays class|

## Creating an Array Explicitly

To **declare** an array.

**syntax**:

```java
datatype[] array_name;
```

Just like with variables, we can **declare and initialize** in the same line.

**syntax**:

```java
datatype[] array_name = {element1, element2, element3, ...}
```

**Example**:

```java
//double array
double[] prices = {13.15, 15.87, 14.22, 16.66};

//String array
String[] clothingItems = {"Tank Top", "Beanie", "Funny Socks", "Corduroys"};
```

## Get Element By Index

Now that we have an array declared and initialized, **we want to be able to get values out of it.**

We use square brackets, **\[** and **\]**, to access data at a certain index:

The index begins with 0

**Example**:

```java
double[] prices = {13.1, 15.87, 14.22, 16.66};

System.out.println(prices[1]);
```

If we try to access an element outside of its appropriate index range, we will receive an **ArrayIndexOutOfBoundsException** error.



```java
double[] prices = {13.1, 15.87, 14.22, 16.66};

System.out.println(prices[5]);
```

## Creating an Empty Array

We can also **create empty arrays** and then fill the items one by one. 

**Empty arrays have to be initialized with a fixed size**.

**syntax**:

```java
datatype[] arrayName = new datatype[num_size]
```

**Example**:

```java
String[] menuItems = new String[5];

//print
System.out.println(Arrays.toString(menuItems))
```

Once you declare this size, it cannot be changed! This array will always be of size 5.

After declaring and initializing, we can set each index of the array to be a different item:

**syntax**:

```java
arrayName[index_num] = some_value;
```

**Example**:

```java
menuItems[0] = "Veggie hot dog";
menuItems[1] = "Potato salad";
menuItems[2] = "Cornbread";
menuItems[3] = "Roasted broccoli";
menuItems[4] = "Coffee ice cream";

//print
System.out.println(Arrays.toString(menuItems))
```

## Length

To get the length of an array, we can access the length attribute of the array object:

**syntax**:

```java
arrayName.length
```

**Example**:

```java
//String[] menuItems = {Veggie hot dog, Potato salad, Cornbread, Roasted broccoli, Coffee ice cream};

System.out.println(menuItems.length);
```

## Array package

The Arrays class in **java.util package** is a part of the Java Collection Framework. 

This class provides static methods to dynamically create and access Java arrays. **It consists of only static methods and the methods of Object class.** The methods of this class can be used by the class name itself.

To import the package in java:

```java
import java.util.Arrays;
```
