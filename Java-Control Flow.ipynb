## Logical Statement

Logical Statement is **a statement result in boolean values (true or false).**

It forms by boolean values and logical operators.

There are 3 logical operators: 
|Logical Operators|Meaning|Description|
|---|---|---|
|**\&\&**|AND|Only true if both A and B are true, otherwise false|
|**\|\|**|OR|Only true if either A or B is true, otherwise false |
|**!**|NOT|negate the booelan value|

## If- Then Statement

**syntax**:

```java
public static void main(String[] args){

    if (logical_statement){
        //execute code if logica_statement is true
    }
    
    //code continue
    ...
}
```

Logical statement returns a **boolean type values**: true/false

**If the logical statement is true, then the block is run.**
**If the logical statement is false, then the block won't run and continue to next code.**

## If-Then Else Statement 

**syntax**:

```java
public static void main(String[] args){

    if (logical_statement){
        //execute code if logica_statement is true
    } else {
        //execute code if logica_statement is false
    }
    
    //code continue
    ...
}
```

**If the logical statement is true, then the if-block is run.**
**If the logical statement is false, then the else-block is run and continue to execute the next code.**

## If-Then-Else-if Statement

**syntax**:

```java
public static void main(String[] args){

    if (logical_statement1){
        //execute code if logical_statement1 is true
    } else if (logical_statement2){
        //execute code if logical_statement2 is true
    } else if (logical_statement3){
        //execute code if logical_statement3 is true
    } else if (...){
        ...
    } else {
        //execute ode if all the logical_statement is false
    } 
    
    //code continue
}
```

If the logical statement1 is true, then the if-block1 is run. If the logical statement is false, then evaluate.

## Nested Conditional Statements

**syntax**:

```java
public static void main(String[] args){

    if (logical_statement1){
        
        //execute code if logical_statement1 is true
        
        if(logical_statement2){
            //execute code if logical_statement1 and logical_statement2 are true
        }
    }
}
```

When we implement nested conditional statements, **the outer statement is evaluated first. If the outer condition is true, then the inner, nested statement is evaluated.** 

## Switch Statement

An alternative to chaining if-then-else conditions together is to use the **switch statement**. This conditional will **check a given value against any number of conditions and run the code block where there is a match.**

**syntax**:

```java
public static void main(String[] args){
    
    var_name = some_value3;
    
    switch (var_name){
        
        //if var_name = some_value1
        case "some_value1":
            //execute this line
            break;
        
        //if var_name = some_value2
        case "some_value2":
            //execute this line
            break;
         
        //if var_name = some_value3
        case "some_value3":
            //execute this line
            break;
        
        //if var_name = some_value4
        case "some_value4":
            //execute this line
            break;   
        ...
        
        //default statement (if none of the var_name matches)
        default:
            //execute this line
    }
}
```

**Example**:

```java
public static void main(String[] args){
    String course = "History";

    switch (course) {
      case "Algebra": 
        System.out.println("Enroll in Algebra");
        break; 

      case "Biology": 
        System.out.println("Enroll in Biology");
        break;

      case "History": 
        System.out.println("Enroll in History");
        break;

      case "Theatre":
        System.out.println("Enroll in Threatre");
        break;

      default:
        System.out.println("Course not found");
    }
}
//Result: Enroll in History
```
