# Comments

```java
A comment is a piece of text within a program that is not executed. It can be used to provide additional information to aid in understanding the code. The # character is used to start a comment and it continues until the end of the line.
```

### Single Line Comment

**syntax**: 

```java
//...
```

**Example**:

```java
//Hello
```

### Double Line Comment

**syntax**: 

```java
/* ... */
```

Example:

```java
/* Hello
Everyone */
```

# Java Code Structure

In order to successfully run a Java code, we simply need 2 structure:
* class variable
* main method

**syntax**:

```java
public class class_name {
    public static void main(String[] args){
        ...Code...;
    }
}
```

Every Java program must have a method called main().

A method is a sequence of tasks for the computer to execute. This main() method holds all of the instructions for our program. 
* All Execuated code must be inside the main method
* every line of code must end with a semicolon ";"

# Variable

In Java, variable needs to be declare first and then assign values.

**List of Java variable primitive datatype:**
|DataType|descriptions|
|---|---|
|**int**| which stores whole numbers||
|**double**| which stores bigger whole numbers and decimal numbers, sufficient for storing 15 decimal digits|
| **boolean**| which stores true and false|
| **char**| which stores a single character/letter or ASCII values using single quotes '...'|
| **String**| which stores multiple characters using double quotes "..."|
| **byte**| Stores whole numbers from -128 to 127|
| **long**| Stores whole numbers from -9,223,372,036,854,775,808 to 9,223,372,036,854,775,807|
| **short**| Stores whole numbers from -32,768 to 32,767|
| **float**| Stores fractional numbers. Sufficient for storing 6 to 7 decimal digits|

In Java, naming the variable must follow a certain way:

1) Variable names of only one word are spelled in all lowercase letters
\
Ex) "**a**"



2) Variable names of more than one word have the first letter lowercase while the beginning letter of each subsequent word is capitalized
\
Ex) "**h**ourly**W**ages" 



3) No other symbols or numbers can begin a variable name
\
Ex) **1**stPlace can't be a variable name


4) A variable starts with \$ or \_ is valid

### To declare variable:

**syntax**:

```java
data_type var_name;
```

**Example**:

```java
//boolean type
boolean isHuman;

//String type
String name;

//integer type 
int age;

//double type
double hourlyWage;

//character type
char sex;
```

### We can also declare multiple variable of the same type

**syntax:**

```java
data_type var_name1,var_name2,var_name3,...
```

**Example**:

```java
int x, y, z;
```

### To assign values to the variable

**syntax**:

```java
var_name = value;
```

Example:

```java
isHuman = true;
name = "Thomas";
age = 20;
hourlyWage = 17.5;
sex = 'm';
```

### We can also assign multiple variable of the same type

```java
var_name1 = var_name2 = var_name3 = ... = some_value;
```

**Example**:

```java
x = y = z = 10;
```

### We can simultaneously declare and assign value to variable

**syntax**:

```java
data_type var_name = some_value;
```

**Example**:

```java
boolean isHuman = true;
String name = "Thomas";
int age = 20;
double hourlyWage = 17.5;
char sex = 'm';
```

### We can simultaneusly declare and assign multiple value to variable

**syntax**:

```java
data_type var_name1, var_name2, var_name3,... = value1;
```

**Example**:

```java
int x, y, z = 10;
```

**or**

```java
int x = 1, y = 2, z = 5;
```

### final Keyword

With the keyword: final placed in front of the declaration, **the value of a variable cannot be changed**

**syntax**:

```java
final datatype var_name = value;
```

**Example**:

```java
final double pi = 3.14;
```

# Variable Manipulation

We are going to go over different kind of variable manipulation:
* Math Operations
* Comparsion Operations
* Compound Assigment Operations
* increment Operations
* String concatenation
* Equality of String

### Math Operations

Basic math operations can be applied to int, double and float data types:

|Math Operators|Descriptions|
|---|---|
|**+**| addition|
|**-**| subtraction|
|**\***| multiplication|
|**/**| division|
|**%**| modulo (yields the remainder)|

These operations are not supported for other data types.

**Example**:

```java
int a = 20;
int b = 10;

int result;

result = a + b;
System.out.println(result);

result = a - b; 
System.out.println(result);

result = a * b; 
System.out.println(result);

result = a / b; 
System.out.println(result);

result = a % b;
System.out.println(result);
```

### Comparison Operators

Comparison operators can be used to compare two values:
|Comparison Operators|Descriptions|
|---|---|
|**>**| greater than|
|**<**|less than|
|**>=**| greater than or equal to|
|**<=**| less than or equal to|
|**==**| equal to|
|**!=**| not equal to|

They are supported for primitive data types and **the result of a comparison is a boolean value true or false**.

Example:

```java
int a = 5;
int b = 3;

boolean result = a > b;
System.out.println(result);

boolean result = a < b;
System.out.println(result);

boolean result = a == b;
System.out.println(result);

boolean result = a != b;
System.out.println(result);
```

### Compound Assignment Operators

Compound assignment operators can be used to change and reassign the value of a variable using one line of code.
|Compound Operators|Descriptions|
|---|---|
|+=|increment by |
|-=|decrement by |
|*=|multiply on top of|
|/=|divide on top of|
|%=|modulo on top of|

Example:

```java
int number = 5;
number += 3;
System.out.println(number);
```

```java
int number = 5;
number -= 4;
System.out.println(number);
```

```java
int number = 5;
number *= 6;
System.out.println(number);
```

```java
int number = 5;
number /= 2;
System.out.println(number);
```

```java
int number = 5;
number %= 7;
System.out.println(number);
```

### Increment and Decrement Operators

The increment operator, (++), can increase the value of a number-based variable by 1 
\
The decrement operator, (--), can decrease the value of a variable by 1.

Example:

```java
int numApples = 5;
numApples++;
System.out.println(numApples);

int numOranges = 5;
numOranges--;
System.out.println(numOranges);
```

### String Concatenation

To concatenate(combine) multiple strings together, we can simply use "+" to do so.

**Example**:

```java
//Concatenate 2 Strings
String str1 = "Hello";
String str2 = "World";
String str3 = str1 + str2;
System.out.println(str3);
```

**We can also concatenate strings and numbers**

Example:

```java
//Concatenate strings and numbers
int num = 5;
System.out.println("I have " + num +" apples.");
```

### Equality of Strings

To compare Strings, the equals() method must be used instead of the primitive equality comparator "=="

**syntax**:

```java
str1.equals(str2);
```

**Example**:

```java
str1 = "apple";
str2 = "apple";
str3 = "Apple";
```

```java
//comparing "apple" and "apple"
str1.equals(str2);
```

```java
//comparing "apple" and "Apple"
str1.equals(str3);
```

**We cannot use "==" operator to check the equality of strings** because
* we use == operators for reference comparison (address comparison) 
* .equals() method for content comparison. 

In simple words, **== checks if both objects point to the same memory location**
\
whereas **.equals() evaluates to the comparison of values in the objects.**
